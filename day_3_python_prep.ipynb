# Q1. Find the factorial of n iteratively — State and handle the edge cases before coding.

```python
'''
Edge cases to be handled
- if n = 0 or 1 then factorial is 1
- if n is negative then factorial is not defined (factorials are defined for positive numbers)
- handle non integers
'''

n = int(input("Enter a number:"))
def factorial(n):
  if n == 0 or n == 1:
    return 1
  elif n < 0:
    return "Factorial is not defined for negative numbers"
  else:
    fact = 1
    while n > 0:
      fact *= n
      n -= 1
    return fact

print(factorial(n))
```

# Q2. Check whether a number is even or odd without converting it to a string — State and handle the edge cases before coding.

```python
'''
Edge cases to be handled
- handle non integers
'''

n = int(input("Enter a number:"))
def even_odd(n):
  if n % 2 == 0:
    return "Even"
  else:
    return "Odd"

print(even_odd(n))
```

# Q3. Reverse a string without using slicing.

```python
string = input("Enter a string:")
reverse_string = ""
for char in string:
  reverse_string = char + reverse_string
print(reverse_string)
```

# Q4. Check whether a string is a palindrome ignoring case.

```python
string = input("Enter a string:")
def is_palindrome(string):
  string = string.lower()
  reverse_string = ""
  for char in string:
    reverse_string = char + reverse_string
  if string == reverse_string:
    return "palendrome"
  else:
    return "not a palendrome"

print(is_palindrome(string))
```
