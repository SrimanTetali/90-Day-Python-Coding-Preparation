# Q1. Check whether a number is even or odd without converting it to a string.

```python
# straight forward approach
num = int(input("Enter a number:"))
def check_even(num):
  if num % 2 == 0:
    print("Even number")
  else:
    print("Odd number")

check_even(num)
```

# Q2. Check whether a number is even or odd without converting it to a string — solve it with a optimized approach.

```python
# optimized method - use bitwise AND
num = int(input("Enter a number:"))

def check_even(num):
  if num & 1 == 0:    # num & 1 gives 0 if even, 1 if odd
    print("Even")     # using the bitwise AND isolates the last bit
  else:
    print("odd")

check_even(num)
```

# Q3. Count the frequency of each character in a string.

```python
text = input("Enter the Text:")
def count_frequency(text):
  freq = {}
  text = text.replace(' ','').lower()
  for i in text:
    freq[i] = freq.get(i,0) + 1
  return freq

FreqCounts = count_frequency(text)
print(FreqCounts)
```

# Q4. Check whether a string is a palindrome ignoring case — First solve it with a straightforward approach, then optimize it.

```python
# straight forward approach

text = input()

def is_palendrome(text):
  text = text.lower()
  rev = ""
  for i in text:
    rev = i + rev
  if rev == text:
    print("Palendrome")
  else:
    print("Not a palendrome")

is_palendrome(text)
```

```python
# another simple way
text = input()

def is_palendrome(text):
  text = text.lower()
  if text == text[::-1]:
    print("Palendrome")
  else:
    print("Not a palendrome")

is_palendrome(text)
```

```python
# optimized way
text = input()

def is_palendrome(text):
  text = text.lower()
  left = 0
  right = len(text) - 1
  while left < right:
    if text[left] != text[right]:
      return False
    left += 1
    right -= 1

  return True

is_palendrome(text)
```

# Q5. Reverse an integer while preserving its sign — Handle duplicate values explicitly in the input.

```python
num = int(input("Enter a number:"))
sign = -1 if num < 0 else 1
num = abs(num)
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

print(rev*sign)
```
