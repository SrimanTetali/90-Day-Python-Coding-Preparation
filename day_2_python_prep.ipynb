# Q1. Find the sum of digits of a positive integer — State and handle the edge cases before coding.

```python
def sum_of_digits(num):
    if not isinstance(num, int):
        return "Input must be integer"

    num = abs(num)

    if num == 0:
        return 0

    total = 0

    while num > 0:
        digit = num % 10
        total += digit
        num //= 10

    return total


print(sum_of_digits(12345))
print(sum_of_digits(0))
```

# Q2. Find the sum of digits of a positive integer — Handle duplicate values explicitly in the input.

handling duplicate values doesn't mean to remove them

```python
def sum_of_digits_with_duplicates(num):
    if not isinstance(num, int):
        return "Input must be integer"

    if num < 0:
        return "Input must be positive integer"

    total = 0

    if num == 0:
        return 0

    while num > 0:
        digit = num % 10
        total += digit
        num //= 10

    return total

print(sum_of_digits_with_duplicates(112233))
print(sum_of_digits_with_duplicates(555))
```

# Q3. Remove duplicate characters while preserving first-seen order — First solve it with a straightforward approach, then optimize it.

```python
def remove_duplicates(text):
    result = ""
    for char in text:
        if char not in result:
            result += char
    return result

print(remove_duplicates("programming"))
print(remove_duplicates("banana"))
```

```python
# optimized approach
def remove_duplicates_optimized(text):
    result = []
    seen = set()
    for char in text:
        if char not in seen:
            result.append(char)
            seen.add(char)
    return "".join(result)

print(remove_duplicates_optimized("programming"))
print(remove_duplicates_optimized("banana"))
```

# Q4. Reverse a string without using slicing — State and handle the edge cases before coding.

```python
def reverse_string(text):
    result = ""
    for i in range(len(text) - 1, -1, -1):
        result += text[i]

    return result

print(reverse_string("hello"))
print(reverse_string("hello world"))
print(reverse_string(""))
print(reverse_string("a"))
```

# Q5. Generate the first n Fibonacci numbers iteratively — First solve it with a straightforward approach, then optimize it.

```python
def fibonacci(n):
    if not isinstance(n, int):
        return "n must be integer"

    if n < 0:
        return "n must be non-negative"

    if n == 0:
        return []

    if n == 1:
        return [0]

    fibonacci = [0, 1]

    for i in range(2, n):
        next_num = fibonacci[i - 1] + fibonacci[i - 2]
        fibonacci.append(next_num)

    return fibonacci


print(fibonacci(5))
print(fibonacci(10))
```

```python
# optimized approach
def fibonacci_optimized(n):
    if not isinstance(n, int):
        return "n must be integer"
    if n < 0:
        return "n must be non-negative"

    result = []
    a = 0
    b = 1
    for _ in range(n):
        result.append(a)
        a, b = b, a + b

    return result

print(fibonacci_optimized(5))
print(fibonacci_optimized(10))
```
